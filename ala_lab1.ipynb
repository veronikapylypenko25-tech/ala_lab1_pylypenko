```python
def plot_2Dshape(X, title, ax=None, color='gray'):
    if ax is None:
        fig,ax=plt.subplots()
    ax.fill(X[0],X[1],color=color, alpha=0.5)
    closed = np.append(X, X[:, [0]], axis=1)
    ax.plot(closed[0],closed[1], color=color)
    ax.axhline(0,color='lightgray', linewidth=0.5)
    ax.axvline(0,color='lightgray', linewidth=0.5)
    ax.set_xlim(-600,600)
    ax.set_ylim(-600,600)
    ax.set_aspect('equal')
    ax.set_title(title)
```

```python
def stretch(X, a, b):
    X_copy=X.copy()
    A=np.array([[a,0], [0,b]])
    X_new= A @ X_copy
    return X_new, A
```

```python
X_str, A_str=stretch(X, 1.5, 0.7)
print(A_str)
plot_2Dshape(X_str, "Stretch (1.5, 0.7)", color='mediumslateblue')
plt.show()
```

```python
def shear(X, a, b):
    X_copy=X.copy()
    A=np.array([[1,a], [b,1]])
    X_new= A @ X_copy
    return X_new, A
```

```python
X_sh, A_sh=shear(X, 0.5, 0)
print(A_sh)
plot_2Dshape(X_sh, "Shear (0.5, 0)", color='lightcoral')
plt.show()
```

```python
def reflection(X, a, b):
    X_copy=X.copy()
    A=np.array([[a**2-b**2,2*a*b], [2*a*b,b**2-a**2]])/(a**2+b**2)
    X_new= A @ X_copy
    return X_new, A
```

```python
X_r, A_r=reflection(X, 1,1)
print(A_r)
plot_2Dshape(X_r, "Reflection (line xy)", color='green')
plt.show()
```

```python
def rotation(X, theta):
    X_copy=X.copy()
    A=np.array([[np.cos(theta), -np.sin(theta)], 
                [np.sin(theta),np.cos(theta)]])
    X_new= A @ X_copy
    return X_new, A
```

```python
def read_off(filename):
    with open(filename, 'r') as f:
        first = f.readline().strip()
        if not first.startswith('OFF'):
            raise ValueError('Not a valid OFF header')
        rest = first[3:].strip()
        if rest:    
            counts = rest.split()
        else:
          counts = f.readline().strip().split()
        n_verts, n_faces = int(counts[0]), int(counts[1])
        verts = [list(map(float, f.readline().split())) for _ in range(n_verts)]
        faces = [list(map(int, f.readline().split()[1:])) for _ in range(n_faces)]
    return np.array(verts), faces
    
vertices, faces = read_off("ModelNet40/airplane/train/airplane_0626.off")
print(vertices.shape)
L = np.linalg.norm(vertices, axis=1).max()
print(L)
```

```python
def plot_3Dshape(V, title, ax=None, color='steelblue'):
    if ax is None:
        fig = plt.figure()
        ax = fig.add_subplot(111, projection='3d')
    ax.scatter(V[:, 0], V[:, 1], V[:, 2], s=1, color=color)
    ax.set_xlim(-L, L)
    ax.set_ylim(-L, L)
    ax.set_zlim(-L, L)
    ax.set_box_aspect((1, 1, 1))
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")
    ax.set_title(title)

plot_3Dshape(vertices, "Original")
plt.show()
```

```python
def rotate_xy(X, theta):
    X_copy = X.copy()
    A = np.array([[np.cos(theta), -np.sin(theta), 0],
                  [np.sin(theta),  np.cos(theta), 0],
                  [0, 0, 1]])
    X_new = (A @ X_copy.T).T
    return X_new, A

def rotate_yz(X, theta):
    X_copy = X.copy()
    A = np.array([[1, 0, 0],
                  [0, np.cos(theta), -np.sin(theta)],
                  [0, np.sin(theta),  np.cos(theta)]])
    X_new = (A @ X_copy.T).T
    return X_new, A

def rotate_xz(X, theta):
    X_copy = X.copy()
    A = np.array([[np.cos(theta), 0, -np.sin(theta)],
                  [0, 1, 0],
                  [np.sin(theta), 0,  np.cos(theta)]])
    X_new = (A @ X_copy.T).T
    return X_new, A
```

```python
def show_pair(V_new, A, title, color):
    print(title)
    print(A)
    fig = plt.figure(figsize=(10, 5))
    ax1 = fig.add_subplot(1, 2, 1, projection='3d')
    ax2 = fig.add_subplot(1, 2, 2, projection='3d')
    plot_3Dshape(vertices, "Original", ax=ax1)
    plot_3Dshape(V_new, title, ax=ax2, color=color)
    plt.show()

V_xy, A_xy = rotate_xy(vertices, np.deg2rad(45))
show_pair(V_xy, A_xy, "Rotation xy (45°)", 'crimson')

V_yz, A_yz = rotate_yz(vertices, np.deg2rad(45))
show_pair(V_yz, A_yz, "Rotation yz (45°)", 'seagreen')

V_xz, A_xz = rotate_xz(vertices, np.deg2rad(45))
show_pair(V_xz, A_xz, "Rotation xz (45°)", 'darkorange')
```

```python
V1, A1 = rotate_xy(vertices, np.deg2rad(30))
V2, A2 = rotate_yz(V1, np.deg2rad(45))
V3, A3 = rotate_xz(V2, np.deg2rad(60))

M = A3 @ A2 @ A1
print(M)
print(np.allclose((M @ vertices.T).T, V3))   
show_pair(V3, M, "Composition: xy(30°)->yz(45°)-> xz(60°)", 'purple')
```
